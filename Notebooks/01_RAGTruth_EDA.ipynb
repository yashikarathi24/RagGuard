```python
import pandas as pd
import numpy as np
import json
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 100)
```

```python
sources = pd.read_json("source_info.jsonl", lines=True)
responses = pd.read_json("response.jsonl", lines=True)
```

```python
responses["hallucinated"] = responses["labels"].apply(
    lambda x: 1 if len(x) > 0 else 0
)

responses["hallucinated"].value_counts()
```

```python
responses["model"].value_counts()
```

```python
responses.groupby("model")["hallucinated"].mean().sort_values(ascending=False) * 100
```

```python
responses["split"].value_counts()
```

```python
responses["split"].value_counts(normalize=True) * 100
```

```python
responses["hallucinated"].value_counts().plot(
    kind="bar",
    figsize=(6,4)
)

plt.title("Hallucinated vs Non-Hallucinated Responses")
plt.xlabel("Class")
plt.ylabel("Number of Responses")
plt.xticks([0, 1], ["Not Hallucinated", "Hallucinated"], rotation=0)
plt.show()
```

```python
responses["hallucinated"].value_counts(normalize=True) * 100
```

```python
from collections import Counter

hallucination_types = Counter()

for labels in responses["labels"]:
    for label in labels:
        hallucination_types[label["label_type"]] += 1

hallucination_types
```

```python
type_counts = pd.Series(hallucination_types).sort_values(ascending=False)

type_counts
```

```python
type_counts.plot(
    kind="bar",
    figsize=(8,4)
)

plt.title("Hallucination Types")
plt.xlabel("Hallucination Type")
plt.ylabel("Number of Annotations")
plt.xticks(rotation=30)
plt.show()
```

```python
data = responses.merge(
    sources[["source_id", "task_type"]],
    on="source_id",
    how="left"
)
```

```python
data.groupby("task_type")["hallucinated"].mean().mul(100).sort_values(ascending=False)
```

```python
data.groupby("task_type")["hallucinated"].mean().mul(100).plot(
    kind="bar",
    figsize=(7,4)
)

plt.title("Hallucination Rate by Task Type")
plt.xlabel("Task Type")
plt.ylabel("Hallucination Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
responses["response_length"] = responses["response"].str.split().str.len()
```

```python
responses.groupby("hallucinated")["response_length"].mean()
```

```python
train_sources = set(
    responses.loc[responses["split"] == "train", "source_id"]
)

test_sources = set(
    responses.loc[responses["split"] == "test", "source_id"]
)

overlap = train_sources.intersection(test_sources)

print("Number of overlapping source IDs:", len(overlap))
```
